# This Notebook Aims to Transform the Raw Dataset to a Clean State Ready for Learning Algorithms

## Importing

```python
from pathlib import Path
import sys
root= Path.cwd().parent
sys.path.append(str(root))
from src.clean import one_hot_encoding

import pandas as pd
```

```python
data = pd.read_csv('../data/raw/raw_data.csv')
```

## Handling Missing Values

```python
data.info()
```

Missing values are only present in 'total_bedrooms' column

```python
row_count = len(data)
missing_count = data['total_bedrooms'].isnull().sum()
print(f"{missing_count*100/row_count}% of rows have a missing value in the bedroom count column")
```

Replace these values by the median (since the ratio is very low and the column is right skewed)

```python
data["total_bedrooms"]= data["total_bedrooms"].fillna( data["total_bedrooms"].median())
```

## Encode Non-Numerical Columns

Inthe column 'ocean_proximity' is encoded using using one-hot encoding

```python
new_columns = one_hot_encoding(data['ocean_proximity'])
```

```python
#drop the ocean_proximity column
data = data.drop(columns=['ocean_proximity'])
#add the new variables
data = pd.concat([data, new_columns], axis=1)
```

## Saving the New Dataset

```python
#full cleaned dataset
data.to_csv('../data/processed/clean_data.csv', index=False)
```
