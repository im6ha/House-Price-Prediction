# This Notebook Aims to Use Linear Discriminant Analysis, Quadratic Discriminant Analysis and Naive Bayes

## Importing

```python
from pathlib import Path
import sys
root= Path.cwd().parent
sys.path.append(str(root))
from src.classification import prior_probabilities, means_estimate, sample_sizes, covariance_matrices, pooled_covariance_matrix, discriminant_analysis_prediction, error_rate, naive_bayes_prediction, parameters_estimate_perclass, henze_zirkler_test
import pandas as pd
```

```python
training_data = pd.read_csv("../data/classification/4_classes/training_data.csv")
testing_data = pd.read_csv("../data/classification/4_classes/testing_data.csv")
training_X= pd.concat([ training_data.iloc[:, 0:3], training_data.iloc[:, 4:]], axis=1 )
training_targets = training_data.iloc[:, 3].to_numpy()

testing_X= pd.concat( [testing_data.iloc[:, 0:3], testing_data.iloc[:, 4:]], axis=1 )
testing_targets = testing_data.iloc[:, 3].to_numpy()
```

## Estimate Prior Probabilities, means and covariance matrices

```python
prior_probas = list(prior_probabilities(training_targets).values())
means = means_estimate(training_X, training_targets).to_numpy()
```

```python
covariance = covariance_matrices(training_X, training_targets, means)
sample = sample_sizes(training_X, training_targets)
pooled_covariance = pooled_covariance_matrix(covariance, sample)
```

## Linear Discriminant Analysis

We assume that X|Y=k follow a multivariate normal distribution with a common covariance matrix for all k's (all classes)

Perform on training dataset

```python
LDA_training_predictions = discriminant_analysis_prediction(training_X, means, pooled_covariance,prior_probas, True)
```

```python
LDA_training_error_rate = error_rate(training_targets, LDA_training_predictions)
```

Perform on testing dataset

```python
LDA_testing_predictions = discriminant_analysis_prediction(testing_X, means, pooled_covariance,prior_probas, True)
```

```python
LDA_testing_error_rate = error_rate(testing_targets, LDA_testing_predictions)
```

```python
print("The Training Error Rate using LDA is: ", LDA_training_error_rate)
print("The Testing Error Rate using LDA is: ", LDA_testing_error_rate)
```

## Quadratic Discriminant Analysis

We assume that X|Y=k follow a multivariate normal distribution

Perform on training dataset

```python
QDA_training_predictions = discriminant_analysis_prediction(training_X, means, covariance,prior_probas, False)
```

```python
QDA_training_error_rate = error_rate(training_targets, QDA_training_predictions )
```

Perform on training dataset

```python
QDA_testing_predictions = discriminant_analysis_prediction(testing_X, means, covariance,prior_probas, False)
```

```python
QDA_testing_error_rate = error_rate(testing_targets, QDA_testing_predictions )
```

```python
print("The Training Error Rate using QDA is: ", QDA_training_error_rate)
print("The Testing Error Rate using QDA is: ", QDA_testing_error_rate)
```

## Naive Bayes

Assume that within each class, the variables (predictors) are independent and assume the distribution of each predictor within every class
    For the Continuous Predictors, assume Normal (Gaussian Distribution)
    For the Binary Predictors (inland	near ocean	island  <1h ocean), assume Bernoulli Distribution

```python
mean_std_estimates = parameters_estimate_perclass(training_X, training_targets)
```

Perform on training dataset

```python
naive_training_predictions = naive_bayes_prediction(training_X, mean_std_estimates, prior_probas)
```

```python
naive_training_error_rate = error_rate(training_targets, naive_training_predictions)
```

Perform on testing dataset

```python
naive_testing_predictions = naive_bayes_prediction(testing_X, mean_std_estimates, prior_probas)
```

```python
naive_testing_error_rate = error_rate(testing_targets, naive_testing_predictions)
```

```python
print("The Training Error Rate using naive bayes is: ", naive_training_error_rate)
print("The Testing Error Rate using naive bayes is: ", naive_testing_error_rate)
```
